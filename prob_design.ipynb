```python
raw_data = "user1:3372,user2:8400,user3:1352" 
# 시간은 초를 나타낸다.

a = raw_data.split(',')

for user in a :
    b,c = user.split(':')
    total = int (c)
    print(b,c)
    d = divmod(total,60)
    #print(d)
    e = divmod(d[0],60)
    # print(e)
    print(f"{b} 의 총 이용시간 - {e[0]}시:{e[1]:02}분:{d[1]}초")
    f = int(c)*10

    if int(c) > 3600:
        f = f * 0.9
    print(f"{b} 의 요금 - {f}원")
    print('='*10)
```
